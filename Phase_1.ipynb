```python
import pandas as pd
import numpy as np
import random

print("=" * 60)
print("PHASE 1: DATA GENERATION (FINAL VERSION - 3500 RECORDS)")
print("=" * 60)

# ---------------------------------------------------------
# STEP 1: GENERATING DATA
# ---------------------------------------------------------
print("\n[Step 1] Generating Synthetic Data with Core Academic & NLP Columns...")
np.random.seed(2026)  # Seed for reproducibility
num_students = 3500   

data = {
    # --- BASIC INFO (UNTOUCHED) ---
    'Student_ID': [], 'Registration_No': [], 'Gender': [], 'Age': [], 
    'Course': [], 'Semester': [], 'Section': [], 'Total_Classes': [], 
    'Classes_Attended': [], 'Attendance_Percentage': [], 
    
    # --- CORE ACADEMIC HISTORY (NEWLY ADDED) ---
    'Previous_Semester_GPA': [], 
    'Current_CGPA': [],
    
    # --- ASSESSMENTS & EXAMS (NEWLY ADDED ASSIGNMENTS) ---
    'Assignment_1': [], 'Assignment_2': [], 'Avg_Assignment_Score': [],
    'Avg_Tutorial_Score': [], 'Avg_Pop_Quiz': [], 'Avg_Lab_Score': [],
    'Mid_Term_Assessment': [], 'Final_Term_Assessment': [], 'Capstone_Project_Marks': [], 
    
    # --- NLP & BEHAVIORAL COLUMNS ---
    'Study_Habits': [],
    'Disciplinary_Record': [],
    'Internship_Status': [],
    'Teacher_Remarks': [],          
    'Extracurricular_Feedback': [], 

    # --- TARGET VARIABLES (HUMAN READABLE) ---
    'Final_Obtained_Score': [], 'Student_Grade_Band': [] 
}

degrees = ['BS Software Eng', 'BS Data Science', 'BS AI', 'BS Info Tech', 'BS Computer Science']

# NLP Text Pools based on Performance Tiers
top_remarks = [
    "Outstanding performance throughout the semester.", 
    "Shows excellent leadership and analytical skills.", 
    "Highly proactive, consistent, and quick learner.", 
    "Brilliant understanding of core concepts. A joy to teach."
]
avg_remarks = [
    "Good effort but needs more consistency in assignments.", 
    "Participates occasionally, can definitely do better.", 
    "Average performance. Needs to focus more on lab work.", 
    "Understands the basics but struggles with complex logic."
]
low_remarks = [
    "Irregular attendance and extremely poor focus in class.", 
    "Needs urgent improvement, counseling recommended.", 
    "Frequently misses deadlines and lacks motivation.", 
    "Lacks fundamental understanding of the subject matter."
]

for i in range(num_students):
    # IDs
    s_id = f'S-{i+10000}' 
    reg_no = f'REG-2026-{i+1000:04d}'
    
    # --- STUDENT LOGIC (To create realistic ML patterns) ---
    student_type = np.random.choice(['Top_Tier', 'Average_Tier', 'Low_Tier'], p=[0.25, 0.55, 0.20])
    
    if student_type == 'Top_Tier':
        att_perc = np.random.randint(85, 100) 
        marks_range = (80, 100)
        gpa_range = (3.5, 4.0)
        t_remark = np.random.choice(top_remarks)
        habit = np.random.choice(["Very consistent", "Regular studier", "Highly focused"])
        disc = "Excellent"
        extra = np.random.choice(["Shows leadership in sports", "Outstanding performer in arts", "Active club president"])
        intern = "Completed"
        
    elif student_type == 'Average_Tier':
        att_perc = np.random.randint(65, 84)
        marks_range = (55, 79)
        gpa_range = (2.5, 3.49)
        t_remark = np.random.choice(avg_remarks)
        habit = np.random.choice(["Occasional studier", "Last-minute crammer", "Needs some guidance"])
        disc = np.random.choice(["Good", "Warning"])
        extra = np.random.choice(["Participates occasionally", "Learning basics of music", "Average sports player"])
        intern = np.random.choice(["Ongoing", "Not Started"])
        
    else: # Low Tier
        att_perc = np.random.randint(30, 64)
        marks_range = (20, 54)
        gpa_range = (1.0, 2.49)
        t_remark = np.random.choice(low_remarks)
        habit = np.random.choice(["Rarely studies", "Last-minute crammer", "Needs constant monitoring"])
        disc = np.random.choice(["Probation", "Needs Improvement", "Warning"])
        extra = np.random.choice(["No interest", "Shy and inactive", "Struggles to participate"])
        intern = "Not Started"

    # Attendance Logic
    total_classes = 50 
    attended = int((att_perc / 100) * total_classes)
    
    # Generate Academic Scores
    prev_gpa = round(random.uniform(gpa_range[0], gpa_range[1]), 2)
    curr_cgpa = round(random.uniform(max(1.0, prev_gpa - 0.3), min(4.0, prev_gpa + 0.3)), 2)
    
    a1 = np.random.randint(marks_range[0], marks_range[1])
    a2 = np.random.randint(marks_range[0], marks_range[1])
    avg_assign = round((a1 + a2) / 2, 2)
    
    avg_tut = np.random.randint(marks_range[0], marks_range[1])
    avg_pq = np.random.randint(marks_range[0], marks_range[1])
    mid = np.random.randint(marks_range[0], marks_range[1])
    final = np.random.randint(marks_range[0], marks_range[1])
    avg_lab = np.random.randint(marks_range[0], marks_range[1])
    capstone = np.random.randint(marks_range[0], marks_range[1])
    
    # Calculate Final Score (Formula mapping to 100%)
    final_score = round(
        (att_perc * 0.05) +          # 5%
        (avg_assign * 0.10) +        # 10%
        (avg_tut * 0.05) +           # 5%
        (avg_pq * 0.10) +            # 10%
        (avg_lab * 0.15) +           # 15%
        (mid * 0.20) +               # 20%
        (final * 0.25) +             # 25%
        (capstone * 0.10), 2         # 10%
    )
    if final_score > 100: final_score = 100.0

    # Assign Human-Readable Grade Band
    if final_score >= 80: band = 'Excellent'
    elif final_score >= 65: band = 'Good'
    elif final_score >= 50: band = 'Average'
    else: band = 'Poor'

    # Append Basic Data
    data['Student_ID'].append(s_id); data['Registration_No'].append(reg_no)
    data['Gender'].append(np.random.choice(['M', 'F']))
    data['Age'].append(np.random.randint(18, 26))
    data['Course'].append(np.random.choice(degrees))
    data['Semester'].append(np.random.randint(1, 9))
    data['Section'].append(np.random.choice(['S1', 'S2', 'S3'])) 
    data['Total_Classes'].append(total_classes); data['Classes_Attended'].append(attended)
    data['Attendance_Percentage'].append(att_perc)
    
    # Append Academic Data
    data['Previous_Semester_GPA'].append(prev_gpa)
    data['Current_CGPA'].append(curr_cgpa)
    data['Assignment_1'].append(a1); data['Assignment_2'].append(a2); data['Avg_Assignment_Score'].append(avg_assign)
    data['Avg_Tutorial_Score'].append(avg_tut)
    data['Avg_Pop_Quiz'].append(avg_pq)
    data['Mid_Term_Assessment'].append(mid)
    data['Final_Term_Assessment'].append(final)
    data['Avg_Lab_Score'].append(avg_lab)
    data['Capstone_Project_Marks'].append(capstone)

    # Append NLP & Text Data
    data['Study_Habits'].append(habit)
    data['Disciplinary_Record'].append(disc)
    data['Internship_Status'].append(intern)
    data['Teacher_Remarks'].append(t_remark)
    data['Extracurricular_Feedback'].append(extra)
    
    # Target
    data['Final_Obtained_Score'].append(final_score); data['Student_Grade_Band'].append(band)

# Convert to DataFrame
df = pd.DataFrame(data)

# ---------------------------------------------------------
# STEP 2: INJECTING NOISE (For Realism)
# ---------------------------------------------------------
print("[Step 2] Injecting Noise to simulate Real World Data...")
raw_df = df.copy()

# Injecting NaNs (Empty cells)
for col in ['Attendance_Percentage', 'Mid_Term_Assessment', 'Teacher_Remarks', 'Current_CGPA']:
    raw_df.loc[raw_df.sample(frac=0.03).index, col] = np.nan 

# Save Raw File
raw_filename = 'raw_dataset.csv'
raw_df.to_csv(raw_filename, index=False)
print(f" > Saved RAW File: '{raw_filename}'")

# ---------------------------------------------------------
# STEP 3: CLEANING DATA 
# ---------------------------------------------------------
print("[Step 3] Cleaning Data for Machine Learning...")
clean_df = pd.read_csv(raw_filename)

# 1. Fill Missing Numeric Values
num_cols = clean_df.select_dtypes(include=[np.number]).columns
for col in num_cols:
    clean_df[col] = clean_df[col].fillna(clean_df[col].mean())

# 2. Fill Missing Text (NLP) Values
cat_cols = clean_df.select_dtypes(include=['object']).columns
for col in cat_cols:
    if clean_df[col].isnull().any():
        if col in ['Teacher_Remarks', 'Extracurricular_Feedback']:
            clean_df[col] = clean_df[col].fillna("No feedback provided")
        else:
            clean_df[col] = clean_df[col].fillna(clean_df[col].mode()[0])

# Clean numeric limits
clean_df['Attendance_Percentage'] = clean_df['Attendance_Percentage'].clip(upper=100)
clean_df['Final_Obtained_Score'] = clean_df['Final_Obtained_Score'].clip(upper=100)
clean_df['Current_CGPA'] = clean_df['Current_CGPA'].clip(upper=4.0)

clean_df = clean_df.round(2)

# Save Clean File
clean_filename = 'clean_dataset.csv'
clean_df.to_csv(clean_filename, index=False)

print(f"✅ DONE! Clean File Saved: '{clean_filename}'")
```
